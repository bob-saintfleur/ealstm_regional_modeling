# Some extra-indications for running
Only two modes have been added to the existing ones provided originally. They are both evaluation-based
- Evaluate on the whole dataset: "evaluate_test" 
- apply the climatology mode: "climatology"

One should start by the "evaluate_test" once the original code philosophy is understood. The output of this mode is required to complete the original camels dataset used in the orchestrator (MLP) we propose. Note that this data is used twice, first for the "lstm_in_mlp" context, secondly for the "predict_lstm_error". Both contexts are detailed in the paper and in the MLP code.

When populating the climatology on the orchestrator, these data are also required. So, make sure this data si ready. The model should use the *maurer_extended* forcing source.

The camels_root/ directory expected should directly hold the folders such as basin_mean_forcing/, usgs_streamflow/, camels_attributes_v2.0/, ......

## 1. Run the evaluate_test mode
Required:
- evaluate_tst -> mode
- camels_root 
- models_box (or run_dir for a single) -> model or dir of models
- nbv -> use only *nbv* first basins, do not use it to include all basins

#### Option 1.1. Run on a single model for all basins

```python
# !python main.py evaluate_test --camels_root path/to/CAMELS --run_dir ./runs/run_2006_0031_seed111 
```

#### Option 1.2. Run on all models in a directory 
Adapt to your reality first

```python
# ! python main.py evaluate_test --camels_root path/to/CAMELS --models_box ./runs 
```

#### Option 1.3. Run on all models in a director on a few basins (use --nbv val)

```python
# ! python main.py evaluate_test --camels_root path/to/CAMELS --models_box ./runs --nbv 2 
```
